```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def __str__(self):
        return f"{self.data}"
    
    def add_node(self, data):
        if self.data > data:
            if self.left is None:
                self.left = Node(data)
            else:
                self.left.add_node(data)

        elif self.data < data:
            if self.right is None:
                self.right = Node(data)
            else:
                self.right.add_node(data)

    def NLR(self):
        print(self.data, end = " ")
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()
    
    def RLN(self):
        if self.right:
            self.right.RLN()
        if self.left:
            self.left.RLN()
        print(self.data, end = " ")

    def RNL(self):
        if self.right:
            self.right.RNL()
        print(self.data, end = " ")
        if self.left:
            self.left.RNL()

    def search(self, number):
        if self.data == number:
            return self
        elif self.data > number:
            if self.left:
                return self.left.search(number)
        elif self.data < number:
            if self.right:
                return self.right.search(number)
        return None
    
    def count_node(self):
        count = 1
        if self.left:
            count += self.left.count_node()
        if self.right:
            count += self.right.count_node()
        return count
    
    def is_empty(self):
        if self.data is None:
            return True
        return False
    
    def leaf_node(self):
        return self.left is None and self.right is None
    
    def one_child_node(self):
        return (self.left and not self.right) or (not self.left and self.right)
    
    def full_node(self):
        return self.left is not None and self.right is not None
    
    def in_chan(self):
        if self.data % 2 == 0:
            print(self.data, end = " ")
        if self.left:
            self.left.in_chan()
        if self.right:
            self.right.in_chan()

    def in_le(self):
        if self.data % 2 != 0:
            print(self.data, end = " ")
        if self.left:
            self.left.in_le()
        if self.right:
            self.right.in_le()

    def in_SNT(self):
        is_prime = True
        if self.data < 2:
            is_prime = False
        else:
            for i in range(2, int(self.data ** 0.5) + 1):
                if self.data % i == 0:
                    is_prime = False
                    break
        
        if is_prime:
            print(self.data, end = " ")
        if self.left:
            self.left.in_SNT()
        if self.right:
            self.right.in_SNT()

    def search_min(self):
        if self.left:
            return self.left.search_min()
        return self.data
    
    def search_max(self):
        if self.right:
            return self.right.search_max()
        return self.data
    
    def sum_node(self):
        total = self.data
        if self.left:
            total += self.left.sum_node()
        if self.right:
            total += self.right.sum_node()
        return total
    
    def avg_node(self):
        return self.sum_node() / self.count_node()
    
    def height(self):
        left_height = 0
        right_height = 0

        if self.left:
            left_height = self.left.height()
        if self.right:
            right_height = self.right.height()
        return max(left_height, right_height) + 1
```

```python
n = Node(12)

n.add_node(7)
n.add_node(62)
n.add_node(61)
n.add_node(90)
n.add_node(5)
n.add_node(13)

print("NLR:")
n.NLR()

print("\nRLN:")
n.RLN()

print("\nRNL:")
n.RNL()


# SEARCH
print("\n\nSearch 61:")
print(n.search(61))

print("Search 100:")
print(n.search(100))


# ĐẾM NODE
print("\nTong node:")
print(n.count_node())


# KIỂM TRA RỖNG
print("\nTree rong:")
print(n.is_empty())


# KIỂM TRA NODE LÁ
print("\nNode la:")
print(n.leaf_node())

print("Node la cua 5:")
print(n.left.left.leaf_node())


# NODE 1 CON
print("\nNode 1 con:")
print(n.right.one_child_node())


# NODE ĐỦ 2 CON
print("\nNode du 2 con:")
print(n.full_node())


# IN SỐ CHẴN
print("\nSo chan:")
n.in_chan()


# IN SỐ LẺ
print("\n\nSo le:")
n.in_le()


# IN SỐ NGUYÊN TỐ
print("\n\nSo nguyen to:")
n.in_SNT()


# MIN - MAX
print("\n\nMin:")
print(n.search_min())

print("Max:")
print(n.search_max())


# TỔNG NODE
print("\nTong gia tri:")
print(n.sum_node())


# TRUNG BÌNH
print("\nTrung binh:")
print(n.avg_node())


# CHIỀU CAO CÂY
print("\nChieu cao:")
print(n.height())
```
