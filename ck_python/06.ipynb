Yêu cầu
Phần 1 — Xây dựng cây

Thêm sản phẩm vào BST theo phần số của product_id, in cây RNL
Thử thêm Product('ABC99', ...) và Product('SP9999', ...) → từ chối nếu mã không hợp lệ
Đếm số node, node lá, node 1 con, node 2 con, chiều cao cây

Phần 2 — Tìm kiếm & thống kê

Tìm sản phẩm theo product_id
Dùng Queue lưu sản phẩm thuộc danh mục Electronics, in theo FIFO
Dùng Stack lưu sản phẩm có giá > 10000000, in theo LIFO
Tính tổng giá trị kho (price * quantity) của toàn bộ sản phẩm
Duyệt LNR in danh sách tăng dần theo product_id

Phần 3 — Cập nhật

Xóa sản phẩm SP012, in lại cây
Cập nhật số lượng SP001 thành 5

Phần 4 — File

Lưu toàn bộ theo LNR ra file products_sorted.txt có tiêu đề cột
Lưu sản phẩm có giá > 10000000 ra file expensive.txt
Lưu sản phẩm có số lượng < 20 ra file low_stock.txt
