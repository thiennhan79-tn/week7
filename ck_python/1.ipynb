```python
class HopDong:
    def __init__(self, mahd, ngayhd, phibh, sotienbh, thoihan):
        self.mahd = mahd
        self.ngayhd = ngayhd
        self.phibh = phibh
        self.sotienbh = sotienbh
        self.thoihan = thoihan

    def __str__(self):
        return f"{self.mahd} - {self.ngayhd} - {self.phibh} - {self.sotienbh} - {self.thoihan}"
```

```python
class Stack:
    def __init__(self):
        self.dsStack = []
    
    def is_empty(self):
        return len(self.dsStack) == 0
    
    def push(self, hd):
        self.dsStack.append(hd)

    def pop(self):
        return self.dsStack.pop()
    
    def __str__(self):
        return "\n".join(str(i) for i in self.dsStack)
```

```python
class Node:
    def __init__(self, data:HopDong):
        self.data = data 
        self.left = None
        self.right = None  
    
    def themhd(self, data:HopDong):
        if self.data.mahd > data.mahd:
            if self.left is None:
                self.left = Node(data)
            else:
                self.left.themhd(data)
        elif self.data.mahd < data.mahd:
            if self.right is None:
                self.right = Node(data)
            else:
                self.right.themhd(data)

    def NLR(self):
        print(self.data)
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()
    
    def read_file(self, path):
        with open(path, "r") as f:
            for line in f:
                row = [row.strip() for row in line.split(",")]
                hd = HopDong(int(row[0]), row[1], int(row[2]), int(row[3]), int(row[4]))
                self.themhd(hd)
    
    def tree(self, prefix="", is_left=True):
        if self.right:
            self.right.tree(prefix + ("│   " if is_left else "    "), False)

        print(prefix + ("└── " if is_left else "┌── ") + str(self.data.mahd))

        if self.left:
            self.left.tree(prefix + ("    " if is_left else "│   "), True)
    
    def search_thoihanLHB3(self, stk=None):
        if stk is None:
            stk = Stack()

        if self.data.thoihan >= 3:
            stk.push(self.data)

        if self.left:
            self.left.search_thoihanLHB3(stk)
        if self.right:
            self.right.search_thoihanLHB3(stk)

        return stk
    
    def write_file_thlhb3(self, path):
        s = self.search_thoihanLHB3()
        with open(path, "w", encoding="utf-8") as f:
           while not s.is_empty():
               hd = s.pop()
               f.write(str(hd) + "\n")
```

```python
hd1 = HopDong(22,'3/7/1965',700000,20000,3)
hd2 = HopDong(69,'28/7/1885',400000,20000,2)
hd5 = HopDong(24,'28/7/1984',400000,20000,2)
hd6 = HopDong(23,'28/7/1984',400000,20000,2)
hd3 = HopDong(13,'10/6/1983',500000,30000,4)
n = Node(hd1)
n.themhd(hd2)
n.themhd(hd3)
n.themhd(hd5)
n.themhd(hd6)
```
