```python
import math
class Node():
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def __str__(self):
        return f"{self.data}"

    def add_node(self, data):
        if self.data > data:
            if self.left is None:
                self.left = Node(data)
            else:
                self.left.add_node(data)

        elif self.data < data:
            if self.right is None:
                self.right = Node(data)
            else:
                self.right.add_node(data)

    def NLR(self):
        print(self.data, end = " ")
        if self.left:
            self.left.NLR()
        if self.right:
            self.right.NLR()
    
    def RLN(self):
        if self.right:
            self.right.RLN()
        if self.left:
            self.left.RLN()
        print(self.data, end = " ")

    def RNL(self):
        if self.right:
            self.right.RNL()
        print(self.data, end = " ")
        if self.left:
            self.left.RNL()
    
    def search(self, number):
        if self.data == number:
            return self
        if self.data > number:
            if self.left:
                return self.left.search(number)
        elif self.data < number:
            if self.right:
                return self.right.search(number)
        return None

    def count(self):
        count = 1
        if self.left:
            count += self.left.count()
        if self.right:
            count += self.right.count()
        return count

    def is_empty(self):
        return self.data is None

    def leaf_node(self):
        return self.left is None and self.right is None  
    
    def node_one_child(self):
        return self.left is None or self.right is None
    
    def full_node(self):
        return self.left and self.right
    
    def sum_node(self):
        sum = self.data
        if self.left:
            sum += self.left.sum_node()
        if self.right:
            sum += self.right.sum_node() 
        return sum
    
    def sum_leaf_node(self):
        if self.left is None and self.right is None:
            return 1
        num = 0
        if self.left:
            num += self.left.sum_leaf_node()
        if self.right:
            num += self.right.sum_leaf_node()
        return num
    
    def node_le(self):
        lst = []
        if self.data is None:
            return None
        if self.data % 2 != 0:
            lst.append(self.data)
        if self.left:
            lst += self.left.node_le()
        if self.right:
            lst += self.right.node_le()
        return lst
    
    def node_chan(self):
        lst = []
        if self.data is None:
            return None
        if self.data % 2 == 0:
            lst.append(self.data)
        if self.left:
            lst += self.left.node_chan()
        if self.right:
            lst += self.right.node_chan()
        return lst
    
    def is_prime(self, n):
        if n < 2:
            return False
        for i in range(2, int(math.sqrt(n)) +1 ):
            if n%i == 0:
                return False
        return True
        
    def node_SNT(self):
        if self.is_prime(self.data):
            print(self.data, end=" ")
        if self.left:
            self.left.node_SNT()
        if self.right:
            self.right.node_SNT()

    def height(self):
        height_left = self.left.height() if self.left else 0
        height_right = self.right.height() if self.right else 0
        return 1 + max(height_left, height_right)
    
    def avg_node(self):
        return self.sum_node()/ self.count()
    
    def min_node(self):
        if self.left is None:
            return self.data
        return self.left.min_node()
    
    def max_node(self):
        if self.right is None:
            return self.data
        return self.right.max_node()
    
    def update_node(self, old_node, new_node):
        if self.data == old_node:
            self.data = new_node
            return
        if old_node < self.data and self.left:
            self.left.update_node(old_node, new_node)
        if old_node > self.data and self.right:
            self.right.update_node(old_node, new_node)

    def delete_node(self, number):
    # TH1: tìm bên trái
        if number < self.data:
            if self.left:
                self.left = self.left.delete_node(number)

        # TH2: tìm bên phải
        elif number > self.data:
            if self.right:
                self.right = self.right.delete_node(number)

        # TH3: node goc
        else:

            # 3.1 Không có  trái
            if self.left is None:
                return self.right

            # 3.2 Không có phải
            if self.right is None:
                return self.left

            # 3.3 Có 2 
            cur = self.right

            while cur.left:
                cur = cur.left

            self.data = cur.data

            self.right = self.right.delete_node(cur.data)

        return self
```
