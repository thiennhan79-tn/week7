Yêu cầu
Phần 1 — Xây dựng cây

Thêm đơn hàng vào BST theo phần số của order_id, in cây RNL
Thử thêm Order('ABC99', ...) và Order('DH9999', ...) → từ chối nếu mã không hợp lệ
Đếm số node, node lá, node 1 con, node 2 con, chiều cao cây

Phần 2 — Tìm kiếm & thống kê

Tìm đơn hàng theo order_id, in thông tin nếu tìm thấy
Dùng Queue lưu đơn hàng có trạng thái Pending, in theo FIFO
Dùng Stack lưu đơn hàng có tổng tiền > 10000000, in theo LIFO
Tính tổng doanh thu từ các đơn Delivered
Duyệt LNR in danh sách tăng dần theo order_id

Phần 3 — Cập nhật

Xóa đơn hàng DH012, in lại cây
Cập nhật trạng thái DH008 thành Delivered

Phần 4 — File

Lưu toàn bộ theo LNR ra file orders_sorted.txt có tiêu đề cột
Lưu đơn hàng Cancelled ra file cancelled.txt
Lưu đơn hàng có tổng tiền > 10000000 ra file high_value.txt
