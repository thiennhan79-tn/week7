Phần 1 — Xây dựng cây

1. Thêm sinh viên vào BST theo phần số của student_id, in cây RNL
2. Thử thêm Student('ABC99', ...) và Student('SV9999', ...) → kiểm tra và từ chối nếu mã không hợp lệ
3. Đếm tổng số node, số node lá, số node 1 con, số node 2 con, chiều cao cây

Phần 2 — Tìm kiếm & thống kê

4. Tìm sinh viên theo student_id, in thông tin nếu tìm thấy
5. Dùng Stack lưu sinh viên có GPA < 2.0, in ra theo thứ tự LIFO
6. Dùng Queue lưu sinh viên ngành CNTT, in theo thứ tự FIFO
7. Duyệt LNR in danh sách sinh viên tăng dần theo student_id

Phần 3 — Cập nhật

8. Xóa sinh viên SV012, in lại cây
9. Cập nhật GPA của SV005 thành 3.9, in lại thông tin node đó
